```python
# if elif else statement for the marks of the students
# 95 and above -> grade A
# 85 to 94 -> grade B
# 75 to 84 -> grade C
# 65 to 74 -> grade D
# below 65 -> grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("You got grade A")
elif 85 <= marks <= 94:
    print("You got grade B")
elif 75 <= marks <= 84:
    print("You got grade C")
elif 65 <= marks <= 74:
    print("You got grade D")
else:
    print("You got grade F")
```
